Indexing and slicing

```python
import numpy as np
```

```python
arr  =  np.arange(1,13)
arr

arr.reshape(4,3)
```

```python
arr[2] ## is return 1st row 3rd elem
```

```python
arr[1:4] # print from 2 to four
```

```python
arr[1:] ### print from 2 o last
```

```python
arr[1:5:2]  ## print 1 to 4 with step size 2
```

```python


arr  =  np.arange(1,13)
my_lists=arr.reshape(4,3)
print(my_lists)
```

```python
# my_lists[0:3] ## print first to 3 rows
```

```python

arr  =  np.arange(1,13)
my_lists=arr.reshape(4,3)
print(my_lists)
```

```python
print(my_lists[0,]) ## return fist row
print(my_lists[1,2]) ## return first row 2nd elem
```

```python
print(my_lists[2:,1:])
```

Boolean Indexing

```python
from numpy import int_
from numpy._core.multiarray import _Array1D


arr: _Array1D[int_] = np.arange(1,26,)
# arr= arr.reshape(5,5) ## Reshape the array into 5 rows and 5
arr
```

```python
arr = np.arange(1,26)

bool_index =  arr %2==0
print(arr[bool_index]) ## print even numbers
```
