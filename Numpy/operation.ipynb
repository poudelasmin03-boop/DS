```python
import numpy as np

arr1  = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])
print(arr1+arr2) ## add two arrays
```

```python
print(arr1*arr2) ## multiply two arrays
```

```python
print(arr1-arr2) ## subtract two arrays
```

```python
print(arr1**arr2)
```

Broadcasting

```python
arr = np.array([1,3,5,7,9])
print(arr+10)
print(arr>3)
print(arr[arr>3]) ## print values greater than 3
```

Deep and shallow copy

```python
arr = np.arange(25)
result = arr.reshape(5,5)
```

```python
mul = result[0:1,0:]
```

```python
print(mul*2)
```
