Array Attributes

```python
arr = np.array(my_list2)
arr.shape ### Give the shape of an array

arr.size ## Give the size of an array

arr.dtype ## Give the data type of an array
```

Array Methods

```python
arr.min()
arr.max()
arr.sum()
```

```python
from typing import Any


from numpy._typing._array_like import NDArray


arr = np.array(my_list2)
print(arr)

row: NDArray[Any] = arr.sum(axis=0) ## sum of each row
print(row)
col = arr.sum(axis=1) ## sum of each column
print(col)

# arr.mean()
```

```python
arr.mean()  ## Mean of an array
arr.std() ## Standard deviation of an array

arr.argmax() ## Index of maximum element
arr.argmin() ## Index of minimum element
```

Reshaping and resizing

```python
arr =  np.arange(1,30,2)
print(arr)


### While try to shape the array the total number of an elements should be same .
## For eg there are 10 elements then array reshape on a [2,5],[5,2],[10,1] or [1,10] but [3,3] is not possible because of there is only 9  elem
np.reshape(arr,(3,5)) ## Reshape an array into 3 rows and 5 columns
```
