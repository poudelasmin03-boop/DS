Matplotlib py

```python
import numpy as np
import  pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,5,11)
x
y=x**2
y
```

```python
plt.plot(x,y)
```

```python
plt.title("Value of x and y")
plt.xlabel("Value of xaxis")
plt.ylabel("Value of  Y axis")
```

Subplot creating a multiple  plot


```python
plt.subplot(2,2,1)
plt.plot(y**2,x**3)
plt.subplot(2,2,2)
plt.plot(x**2,y)
plt.subplot(2,2,3)
plt.plot(y,x)
plt.subplot(2,2,4)
plt.plot(x,y)
```

object oritend way 

```python
fig = plt.figure(figsize=(12,8))
### Left bottom size width height values always btn 0 and 1
axis = fig.add_axes([0.1,0.1,0.8,0.8])
axis.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,0.8,0.8])
axis2.plot(x**2,y)
```

```python
fig  = plt.figure()
axis = fig.add_axes([0.1,0.1,0.8,0.8])
axis.plot(x,y)
plt.savefig("Plot.jpg")
```
