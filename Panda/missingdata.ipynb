Missing data 

```python
import numpy as np
import pandas as pd
```

```python
data = {
    "A":[1,2,3,4,5],
    "B":[1,2,3,4,5],
    "C":[1,2,3,np.nan,np.nan],
    "D":[1,np.nan,np.nan,np.nan,5]
}
df = pd.DataFrame(data)
df
```

```python
### Find the nan in true and false then sum how  mmay nan in column
df.isna().sum()
```

```python
## Check the null value in Column
df.isna().any()
```

```python
## Remove Nan it drop null value row
df.dropna()
```

```python
df.dropna(thresh=3)
```

Filling missing data

```python
df
```

```python
df.fillna(0)
```

```python
value = {
    "C":300,
    "D":100
}
```

```python
#### Fill column wise
df.fillna(value)
```

```python
df.fillna(df.mean())
```
