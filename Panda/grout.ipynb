Group by Aggragation

```python
import numpy as np
import pandas as pd
```

```python
from pandas import date_range


data = {
   "Category":["A","B","A","B","A"],
   "Store":["S1","s2","s1","S2","S1"],
   "Sales":[100,200,32,102,200],
   "Quantity":[10,12,23,12,43],
   "Date":pd.date_range('2082-1-02',periods=5)
}
df = pd.DataFrame(data)
```

```python
group = df.groupby("Category")["Sales"].sum()
group
```

```python
group = df.groupby("Store")["Sales"].sum()
group
```

```python
##multiple column
group = df.groupby(['Category','Store'])["Sales"].sum()
group
```
