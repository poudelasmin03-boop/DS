Merging and concat

```python
import numpy as np
import pandas as pd
```

```python
employees = {
    "employee_id":[1,2,3,4,5],
    "name":["jhon","Asmin","Nitik","Ramesh","Shaym"],
    "Program":["Go","Python","PHP","Rust","R"],

}

Salaries = {
    "employee_id":[1,2,3,5,7],
    "salary":[60000,70000,90000,30000,200000],
    "bonus":[50000,10000,80000,120000,20000]
}


df1 = pd.DataFrame(employees)
df2 =pd.DataFrame(Salaries)
df1
```

```python
df2
```

```python
pd.merge(df1,df2,on="employee_id",how="inner")
```

```python
pd.merge(df1,df2,on="employee_id",how="outer")
```
