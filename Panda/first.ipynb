```python
import pandas as pd
import numpy as np
print(pd.__file__)
print(pd.__version__)
print(hasattr(pd, "Series"))
```

Series is a 1d labeled array capable or holding any data type

```python
labels = ['a','b','c']
my_list = [10,20,30]
arr = np.array([10,20,30])
d={
    "name":"Asmin Poudel",
    "age":21
}
print(pd.Series(my_list, index=labels))
```

```python
#### Only one 1d array
print(pd.Series(arr, index=labels))
```

```python
print(pd.Series(d, index=labels))
```
