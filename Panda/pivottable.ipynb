Pivot Table

```python
import pandas as pd
import numpy as np
```

```python

import random

data = {
    "Date": [
        f"2026-{random.randint(1, 12):02d}-{random.randint(1, 28):02d}"
        for _ in range(1000)
    ],

    "Product": [
        random.choice([
            "Laptop",
            "Mobile",
            "Tablet",
            "Headphones",
            "Keyboard",
            "Mouse",
            "Monitor",
            "Printer",
            "Smartwatch",
            "Camera"
        ])
        for _ in range(1000)
    ],

    "Region": [
        random.choice([
            "North",
            "South",
            "East",
            "West",
            "Central"
        ])
        for _ in range(1000)
    ],

    "Sales": [
        round(random.uniform(500, 100000), 2)
        for _ in range(1000)
    ],

    "Units": [
        random.randint(1, 100)
        for _ in range(1000)
    ]
}

df = pd.DataFrame(data)

# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])

print(df)
```

```python
### Check null value at first
df.isna().sum()
```

```python
pd.pivot_table(df,values="Sales",index="Region",columns="Product",aggfunc="median")
```

```python
pd.pivot_table(df,values=["Sales","Units"],index="Region",columns="Product",aggfunc="median")
```
